```python
from firecrawl import Firecrawl
from dotenv import load_dotenv
import os
import json
from pydantic import BaseModel
from typing import List

load_dotenv()
```

```python
app = Firecrawl(api_key=os.getenv("firecrawl_api_key"))

# Autoren, die in Niederlande geboren sind (24.836) und als Tabelle dargestellt (10 results per page)
url_1 = "https://www.dbnl.org/zoek/solr/auteurZoek.php?zoekURLKey=6ce29d1358509682f2254bc472fad1e7&weergave=tabel&p=1"

# Autoren, die in Niederlande geboren sind (24.836) und als Tabelle dargestellt (100 results per page)
url_2 = "https://www.dbnl.org/zoek/solr/auteurZoek.php?zoek=&prefilter=auteurs&subform_values=showfacet_geslacht%3D1%26geslacht%3D%26showfacet_periode_g%3D1%26showfacet_periode_o%3D1%26showfacet_landtype_g%3D1%26landtype_g%255B%255D%3Dneder01%26showfacet_landtype_o%3D1%26showfacet_deelcollectie%3D1%26showfacet_rol%3D1&f_id=list_size_subform&size=100&sort=snaam%2Basc&weergave=tabel"
```

```python
# 1) Scrape webpage with json output

## define class
class AuthorsTable(BaseModel):
    author_name: list[str]
    author_profile_url: list[str]
    birth_year: list[str]
    death_year: list[str]


result = app.scrape(
    url_1,
    formats = [{
        "type": "json",
        "schema": AuthorsTable.model_json_schema(),
        "prompt": "Extract the author name (NAAM), the url to the author profile, the birth year (GEBOORTEJAAR) and the death year (STERFJAAR) of each author in the table."
    }]
)


print(json.dumps(result.json, indent=4, ensure_ascii=False))
```

- scrape was complete and all scraped data was correct

```python
# 1. Define the schema for a single Author
class Author(BaseModel):
    author_name: str
    author_profile_url: str
    birth_year: str
    death_year: str

# 2. Define a wrapper if the tool requires a JSON object 
class AuthorsList(BaseModel):
    authors: List[Author]


# web page with 10 authors in the table
result = app.scrape(
    url_1,
    formats=[{
        "type": "json",
        # Use the list-based schema
        "schema": AuthorsList.model_json_schema(), 
        "prompt": "Extract the author name (NAAM), the url to the author profile, the birth year (GEBOORTEJAAR) and the death year (STERFJAAR) of each author in the table."
    }]
)

print(json.dumps(result.json, indent=4, ensure_ascii=False))
```
